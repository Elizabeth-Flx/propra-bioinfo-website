```python
import matplotlib.pyplot as plt

with open("riddle2.txt", "r") as f:
    x = []
    y = []

    for line in f:
        stripped = line.strip()

        if stripped:
            split = stripped.split(" ")
            x.append(float(split[1]))
            y.append(float(split[2]))
        else:
            plt.plot(x, y)
            x.clear()
            y.clear()
    if x and y:
        plt.plot(x, y)

plt.savefig("plot.png")
```
