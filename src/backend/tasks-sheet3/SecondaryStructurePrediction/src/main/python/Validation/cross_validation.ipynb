```python
from wrapper import cross_validate
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import os
```

```python
out_dir = "/home/nico/Software/propra/src/backend/tasks-sheet3/SecondaryStructurePrediction/src/main/python/Validation/plots/k_fold"
database = "/home/nico/Software/propra/src/backend/tasks-sheet3/SecondaryStructurePrediction/data/CB513DSSP.db"
sequences = "/home/nico/Software/propra/src/backend/tasks-sheet3/SecondaryStructurePrediction/data/CB513.fasta"
```

```python
def compare_versions(database: str, sequences: str, k: int = 5):
    versions = ["gor1", "gor3", "gor4"]

    results_per_version = {}

    for version in versions:
        result = cross_validate(database, sequences, version, k=k)

        results_per_version[version] = result

    return results_per_version
```

```python
res = compare_versions(database, sequences)
```

```python
versions = ["gor1", "gor3", "gor4"]

for metric in res["gor1"]:
    df = pd.DataFrame([res[version][metric] for version in versions]).T
    df.columns = versions
    sns.violinplot(data=df)
    plt.title(metric, fontsize=30)
    plt.xticks(fontsize=20)
    plt.yticks(fontsize=20)
    plt.ylim(0, 100)
    path = os.path.join(out_dir, f"{metric}.png")
    plt.savefig(path)
    plt.clf()
```
