```python
input_database = "PDB_40/pdb40.db"
input_filtered = "PDB_40/pdb40_filtered.fasta"
output_database = "PDB_40/pdb40_filtered.db"
```

```python
with open(input_database, "r") as f:
    database = f.read()

with open(input_filtered, "r") as f:
    filtered = f.readlines()
```

```python
ids = set([line.strip() for line in filtered if line.startswith(">")])
len(ids)
```

```python
blocks = [block for block in database.split("\n\n") if len(block.splitlines()) > 1 and block.splitlines()[0] in ids]
len(blocks)
```

```python
with open(output_database, "w+") as f:
    for block in blocks:
        f.write(block + "\n\n")
```
