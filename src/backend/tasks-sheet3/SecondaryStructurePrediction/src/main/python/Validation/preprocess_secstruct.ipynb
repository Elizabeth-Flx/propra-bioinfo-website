```python
import os
```

```python
raw_dir = "pdb40.raw"
output_db = "pdb40.db"
output_fasta = "pdb40.fasta"
```

```python
with open(output_db, "w+") as db:
    with open(output_fasta, "w+") as fasta:
        for file in os.listdir(raw_dir):
            with open(os.path.join(raw_dir, file), "r") as f:
                lines = f.readlines()
                id = lines[0].split(": ")[0].strip()
                sequence = lines[0].split(": ")[1].strip()
                secondary = lines[3].split(": ")[1].upper().replace(" ", "C").strip()

                if len(sequence) < 5:
                    continue

                db.write(f"> {id}\nAS {sequence}\nSS {secondary}\n\n")
                fasta.write(f"> {id}\n{sequence}\n\n")
```
