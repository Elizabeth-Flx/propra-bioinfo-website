```python
from wrapper import evaluate, combine_dicts
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
out_dir = "/home/nico/Software/propra/src/backend/tasks-sheet3/SecondaryStructurePrediction/src/main/python/Validation/plots/test"
training = "/home/nico/Software/propra/src/backend/tasks-sheet3/SecondaryStructurePrediction/cb513.db"
validation = "/home/nico/Software/propra/src/backend/tasks-sheet3/SecondaryStructurePrediction/src/main/python/Validation/PDB_40/pdb40_filtered.db"
sequences = "/home/nico/Software/propra/src/backend/tasks-sheet3/SecondaryStructurePrediction/src/main/python/Validation/PDB_40/pdb40_filtered.fasta"
os.makedirs(out_dir, exist_ok=True)
```

```python
results = {}

for version in ["gor1", "gor3", "gor4"]:
    raw = evaluate(training, sequences, validation, version, "pdb")
    results[version] = combine_dicts([raw])
```

```python
versions = ["gor1", "gor3", "gor4"]

for metric in results["gor1"]:
    df = pd.DataFrame([results[version][metric] for version in versions]).T
    df.columns = versions
    sns.violinplot(data=df)
    plt.title(metric, fontsize=30)
    plt.xticks(fontsize=20)
    plt.yticks(fontsize=20)
    plt.ylim(0, 100)
    path = os.path.join(out_dir, f"{metric}.png")
    plt.savefig(path)
    plt.clf()
```
